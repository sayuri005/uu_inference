```python
import numpy as np
import pandas as pd 
import random
```

```python
#7
prices_data = pd.read_csv('./data/ames_prices.csv')
prices_data["Central.Air"].head()
```

```python
for i in range(len(prices_data['Central.Air'])):
    if prices_data.loc[i,'Central.Air'] == 'Y':
        prices_data.at[i,'Central.Air'] = '1'
    else:
        prices_data.at[i,'Central.Air'] = '0'
```

```python
prices_data['Central.Air']=prices_data['Central.Air'].astype(int)
```

```python
#a
y_central_air=prices_data['Central.Air'].mean()
y_central_air
```

```python
#b
R = 3000
rng = np.random.default_rng(50)
y_central_air_b = []

for t in range(R):

    sample_b = prices_data['Central.Air'].sample(frac=1.0, replace=True, random_state=rng)
   
    prop_b = sample_b.mean() 

    y_central_air_b.append(prop_b)
  
```

3000 replication and seed=50

```python
#c
air_se = np.std(y_central_air_b,ddof=1)
air_se
```

```python
#d
ci90_low = y_central_air - 1.645*air_se
ci90_high = y_central_air + 1.645*air_se

ci95_low = y_central_air - 1.96*air_se
ci95_high = y_central_air + 1.96*air_se


ci99_low = y_central_air - 2.576*air_se
ci99_high = y_central_air + 2.576*air_se


print(f'90%: ({ci90_low},{ci90_high}), 95%: ({ci95_low},{ci95_high}),99%: ({ci99_low},{ci99_high})')
```

e. 
At significance level 0.1, we reject the null hypothesis because 0.9 is not in its respective interval. At significance level 0.05, we reject the null hypothesis because 0.9 is not in its respective interval. At significance level 0.01, we reject the null hypothesis because 0.9 is not in its respective interval. This suggests that the homes with central air is not 90% of all the homes. Looking at the confidence intervals, it suggests that the proportion may be higher. 

```python
#8 
prices_corr = prices_data[['area','price']]
prices_corr_arc = np.arcsinh(prices_corr)

initial_corr = prices_corr_arc['area'].corr(prices_corr_arc['price'])
initial_corr
```

```python
#a
R = 3000
rng2 = np.random.default_rng(50)
area_price_corr_b = []

for t in range(R):

    sample_price_area = prices_corr_arc.sample(frac=1.0, replace=True, random_state=rng2)
    corr_b = sample_price_area['area'].corr(sample_price_area['price'])

    area_price_corr_b.append(corr_b)
```

3000 replication and seed=50

```python
#b
corr_se = np.std(area_price_corr_b,ddof=1)
corr_se
```

```python
#c
ci90_low_corr = initial_corr - 1.645*corr_se
ci90_high_corr = initial_corr + 1.645*corr_se

ci95_low_corr = initial_corr - 1.96*corr_se
ci95_high_corr = initial_corr + 1.96*corr_se


ci99_low_corr = initial_corr - 2.576*corr_se
ci99_high_corr = initial_corr + 2.576*corr_se


print(f'90%:({ci90_low_corr},{ci90_high_corr}), 95%: ({ci95_low_corr},{ci95_high_corr}) ,99%:({ci99_low_corr},{ci99_high_corr})')
```

d. 
At significance level 0.1, we reject the null hypothesis because 0 is not in its respective interval. At significance level 0.05, we reject the null hypothesis because 0 is not in its respective interval. At significance level 0.01, we reject the null hypothesis because 0 is not in its respective interval. This suggests that a homes' area and price are correlated with one another - an increase in area means an increase in price of the home (as the interval values are positive not negative). 

e. 
Area and price must remain paired during resampling as the variables are dependent on one another. Sampling them separately will fail to take into account their dependence and lead to misleading bootstrap values. 

```python
#9
survival_data = pd.read_csv('./data/metabric.csv')
survival_data.head()
```

```python
initial_survival  = survival_data['Overall Survival (Months)'].mean()
initial_survival
```

```python
#a
R = 3000
rng3 = np.random.default_rng(50)

mean_b =[]

for t in range(R):

    sample_survival = survival_data['Overall Survival (Months)'].sample(frac=1.0, replace=True, random_state=rng3)
    sample_survival_mean_b = sample_survival.mean()

    mean_b.append(sample_survival_mean_b)
```

```python
#b
survival_se = np.std(mean_b,ddof=1)
survival_se
```

```python
#c
ci90_low_survival = initial_survival - 1.645*survival_se
ci90_high_survival = initial_survival + 1.645*survival_se

ci95_low_survival = initial_survival - 1.96*survival_se
ci95_high_survival = initial_survival + 1.96*survival_se


ci99_low_survival = initial_survival - 2.576*survival_se
ci99_high_survival = initial_survival + 2.576*survival_se


print(f'90%: ({ci90_low_survival},{ci90_high_survival}),95%: ({ci95_low_survival},{ci95_high_survival}), 99%: ({ci99_low_survival},{ci99_high_survival})')
```

d. At significance level 0.1, we reject the null hypothesis because 120 is not in its respective interval. At significance level 0.05, we reject the null hypothesis because 120 is not in its respective interval. At significance level 0.01, we reject the null hypothesis because 120 is not in its respective interval.  


e.This means that the mean survival months or date of last follow up is not 120 months or 10 years as we rejcted the null for all three significance levels. The sample mean suggests that the mean survival months or date of last follow up is around 129 months or 10.75 years and the confidence interval shows the interval where the true value may lie (the percentage of time it will be in the confidence interval depends on the confidence level). The results do not estimate unrestricted mean lifetime as it cuts off the survival months for patients who were living at the time when the last follow up was recorded. The mean underestimates unrestricted mean lifetime as the value of "months of survival" for those who continued to live would have been greater than what was recorded in the data set.  

```python
#10
sleep_data = pd.read_csv('./data/nhanes_data_17_18.csv')
```

```python
sleep_data_two=sleep_data[['AgeInYearsAtScreening','SleepHoursWeekdaysOrWorkdays']]
sleep_data_no_na=sleep_data_two.dropna()
sleep_data_18=sleep_data_no_na[sleep_data_no_na['AgeInYearsAtScreening']>=18]
```

```python
initial_sleep = sleep_data_18['SleepHoursWeekdaysOrWorkdays'].mean()
initial_sleep
```

```python
#a
R = 3000
rng = np.random.default_rng(50)
sleep_all = []

for t in range(R):

    sleep_b = sleep_data_18['SleepHoursWeekdaysOrWorkdays'].sample(frac=1.0, replace=True, random_state=rng)
   
    sleep_mean_b = sleep_b.mean() 

    sleep_all.append(sleep_mean_b)
  
```

Seed: 50 and number of replications: 3000 

```python
#b
sleep_se = np.std(sleep_all,ddof=1)
sleep_se
```

```python
#c
ci90_low_sleep = initial_sleep  - 1.645*sleep_se
ci90_high_sleep  = initial_sleep  + 1.645*sleep_se

ci95_low_sleep  = initial_sleep  - 1.96*sleep_se
ci95_high_sleep  = initial_sleep  + 1.96*sleep_se


ci99_low_sleep  = initial_sleep  - 2.576*sleep_se
ci99_high_sleep  = initial_sleep  + 2.576*sleep_se


print(f'90%: ({ci90_low_sleep},{ci90_high_sleep}) ,95%: ({ci95_low_sleep},{ci95_high_sleep}), 99%: ({ci99_low_sleep},{ci99_high_sleep})')
```

d. At significance level 0.1, we reject the null hypothesis because 8 is not in its respective interval. At significance level 0.05, we reject the null hypothesis because 8 is not in its respective interval. At significance level 0.01, we reject the null hypothesis because 8 is not in its respective interval.  

e. The mean hours of sleep on workdays/weekdays of people who are represented in the survey, who are 18 or older, is not 8 hours. The confidence interval suggests that the number of hours of sleep is lower than 8 hours. Practical and statistical significance do not tell the same story as although we reject the null that the mean is 8, the confidence interval is giving numbers values that are less than 8 but the difference seems to be between 0.5 and 0.7 which is not a huge difference in sleep hours.
